# Working with gridded spatial data in R

## Overview 

In this notebook, we will be introduced to packages for working with gridded spatial data in R. We will work with multidimensional gridded data using `stars` (Spatiotemporal Arrays) and perform high-performance raster operations using `terra`.

### Key Library Adjustments
| Feature |	Python Library | R Library
|---|---|---|
|N-Dimensional Arrays Analysis |	xarray	| stars
|N-Dimensional Arrays Storage |	zarr	| zarr / Rarr / pizzarr
|Raster Manipulation |	rioxarray / rasterio | terra
|NetCDF/HDF5 Handling |	netCDF4	| ncdf4

### 1. Load the necessary libraries

```python
# Install packages if necessary
install.packages(c("terra", "stars", "aws.s3", "abind", "progressr"))

# Connecting to data on AWS (equivalent to Python 'fsspec')
library(aws.s3) 

# For gridded data (equivalent to xarray)
library(stars) 
library(terra) # Alternative for raster/gridded data

# For arrays (equivalent to numpy - base R handles arrays naturally)
# No specific library needed, but 'abind' is useful for manipulation
library(abind)

# Progress bars (equivalent to dask.diagnostics)
library(progressr) 
handlers(global = TRUE)
```

### 2. Gridded data with Xarray

```python
# Don't print warnings (equivalent to warnings.filterwarnings)
# options(warn = -1)

# Create a 2x3 matrix
# byrow=TRUE ensures it matches the Python row-major layout
arr <- matrix(c(1, 2, 3, 5, 6, 7), nrow = 2, byrow = TRUE)

# Create a stars object by defining dimensions and coordinates
xda <- st_as_stars(arr)
xda <- st_set_dimensions(xda, names = c("x", "y"))
xda <- st_set_dimensions(xda, "x", values = c(10, 20))
xda <- st_set_dimensions(xda, "y", values = c(1.1, 1.2, 1.3))

xda
```

We can access the individual components like the data itself, the dimension names or the coordinates using accessors

```python
# Get the actual data
print(as.matrix(xda))

# Get the dimension names
print(names(st_dimensions(xda)))

# Get the x coordinates
print(st_dimensions(xda)$x$values)
```

We can set or get any metadata attribute we like

```python
# Add attributes
attr(xda, "long_name") <- "random measurement"
attr(xda, "random_attribute") <- 123

print(attributes(xda))
```

and perform calculations on `xarray.DataArrays` as if they were Numpy arrays

```python
# Arithmetic and functions
xda + 10
sin(xda)
```
